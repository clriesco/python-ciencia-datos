### 👀 Demo F6.1 · Definir, llamar, devolver

Con la tabla entera, `es_menor` falla: hay un `NA`.

Otra función, ya escrita: un parámetro con **valor por defecto** (`decimales=2`) y una llamada con el **argumento por su nombre** (*keyword argument*). Leedla y ejecutadla.

````python
def tasa(lista, decimales=2):
    """Calcula la fracción de unos en una lista de 0 y 1.

    Args:
        lista: Los valores, 0 o 1.
        decimales: Con cuántos decimales se redondea. Por defecto, 2.

    Returns:
        La fracción, redondeada.
    """
    return round(sum(lista) / len(lista), decimales)


print(tasa([1, 0, 1]))                    # decimales toma su valor por defecto
print(tasa([1, 0, 1], decimales=1))       # o se le da, por su nombre
````
