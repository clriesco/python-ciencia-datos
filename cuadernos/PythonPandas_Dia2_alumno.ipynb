````python
# Cargar los datos de Covid en España (una fila por día, cifras acumuladas) y el resumen por países.
# Ejecutad esta celda sin cambiar nada.
import pandas as pd

confirmados = pd.read_csv("https://raw.githubusercontent.com/clriesco/python-ciencia-datos/main/data/covid_confirmed.csv")
fallecidos = pd.read_csv("https://raw.githubusercontent.com/clriesco/python-ciencia-datos/main/data/covid_deaths.csv")
recuperados = pd.read_csv("https://raw.githubusercontent.com/clriesco/python-ciencia-datos/main/data/covid_recovered.csv")
resumen = pd.read_csv("https://raw.githubusercontent.com/clriesco/python-ciencia-datos/main/data/covid_summary.csv")
print(confirmados.shape, fallecidos.shape, recuperados.shape, resumen.shape)
````

### 🎲 Apuesta 6.1 · Lo que no empareja

`resumen` tiene 197 países; `poblacion`, 5 (en millones, cifras aproximadas de 2020). ¿Cuántas filas tiene el resultado de `merge`?

*Antes de ejecutar, escribid vuestra apuesta en la primera línea de la celda.*

````python
# Mi apuesta: 

poblacion = pd.DataFrame({"Country": ["Spain", "Italy", "France", "Germany", "Portugal"],      # una tabla a mano: un diccionario de columnas
                          "Population": [47.4, 59.6, 67.4, 83.2, 10.3]})
unido = resumen[["Country", "TotalConfirmed", "TotalDeaths"]].merge(poblacion, on="Country")
print(len(unido))
````

> 📽️ *Mirad la pantalla: diapositiva 37 · «Fechas».*

### 👀 Demo 6.3 · Fechas

`Date` llega como texto. `to_datetime` lo convierte en fecha, y una fecha sabe su año, su mes y su día: `.dt`.

Con `.dt` se saca el año, el mes, el día de la semana; y se compara:

### 👀 Demo 6.4 · De acumulado a diario: diff

Las cifras del fichero son acumuladas: cada día, el total desde el principio. `diff` resta a cada fila la anterior.

````python
# La tabla combinada y con fechas, por si os habéis perdido algún paso. Ejecutad esta celda sin cambiar nada.
covid = (confirmados[["Date", "Confirmed"]]
         .merge(fallecidos[["Date", "Deaths"]], on="Date")
         .merge(recuperados[["Date", "Recovered"]], on="Date"))
covid["Date"] = pd.to_datetime(covid["Date"].str[:10])
covid["NewConfirmed"] = covid["Confirmed"].diff()
covid["NewDeaths"] = covid["Deaths"].diff()
print(covid.shape)
````

### 🎲 Apuesta 6.2 · Resumir por mes

`dt.strftime("%Y-%m")` convierte cada fecha en un texto como `2020-04`. Con lo que sabéis de `groupby`, ¿qué devuelve esta línea, y qué mes creéis que sale primero?

*Antes de ejecutar, escribid vuestra apuesta en la primera línea de la celda.*

````python
# Mi apuesta: 

covid.groupby(covid["Date"].dt.strftime("%Y-%m"))["NewDeaths"].sum().sort_values(ascending=False).head(3)
````
