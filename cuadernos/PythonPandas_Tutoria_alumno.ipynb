### 🐞 Ejercicio A.7 · Media de columnas de texto · cazad el error

### 🐞 Ejercicio A.8 · Un entero con huecos · cazad el error

Tiene que crear la columna `AgeInt` con la edad redondeada como entero.


````python
df["AgeInt"] = df["Age"].round().astype(int)
print(df["AgeInt"].head(3))
````

### 🐞 Ejercicio A.9 · No falla, pero no ordena · cazad el error

### 🐞 Ejercicio A.10 · No falla, pero miente · cazad el error

Esta celda quiere la proporción de menores de 18 **entre los pasajeros con edad conocida**, y da 0.07. La cifra correcta es 0.13. Encontrad el fallo.

**Debe salir:**
```text
0.13
```

````python
print(round((df["Age"] < 18).mean(), 2))
````
